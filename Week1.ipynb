What is the most popular female name on the ship? Extract the passenger’s full name
(Name column) from his name (First Name). This task is a typical example of what a data
analysis specialist is faced with. The data is very heterogeneous and noisy, but you need to
extract the necessary information from it. Try to manually parse several values of the Name
column and work out a rule for extracting names, as well as dividing them into female and
male.

```python
def extract_gender_and_first_name(name):
    # Split last name, rest
    parts = name.split(', ')
    if len(parts) > 1:
        rest = parts[1]  # "Title. First Name(s)"
        title, *first_names = rest.split(' ')
        if title in ['Miss.', 'Mrs.']:
            gender = 'female'
        else:
            gender = 'male'
        # join first names as string
        first_name = ' '.join(first_names)
        return gender, first_name
    else:
        return None, None
```

```python
Titanic['Gender_Name'] = Titanic['Name'].apply(lambda x: extract_gender_and_first_name(x))
Titanic['Gender'] = Titanic['Gender_Name'].apply(lambda x: x[0])
Titanic['First_Name'] = Titanic['Gender_Name'].apply(lambda x: x[1])
```
