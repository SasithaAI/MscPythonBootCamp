### What Part of passangers managed to Survive ?

```python
Survival_Precentage= format((Titanic['Survived'].value_counts().values[1]/Titanic['Survived'].value_counts().sum())*100)
Survival_Precentage = round(float(Survival_Precentage),2)
print(f" The Survival Precentage is : {Survival_Precentage} %")
```

### What is the Proportion of First-Class Passangers?

```python
Titanic['Pclass'].value_counts()
```

```python
FirstClass_Precentage= format((Titanic['Pclass'].value_counts().values[1]/Titanic['Pclass'].value_counts().sum())*100)
FirstClass_Precentage = round(float(FirstClass_Precentage),2)
print(f" The First Class Passangers Precentage is : {FirstClass_Precentage} %")
```

### How old Were the Passangers?

```python
Mean_Age = Titanic['Age'].mean()
Median_Age = Titanic['Age'].median()
print(f"The Mean Age is : {Mean_Age}"+" "+f"The Median Age is : {Median_Age} ")
```

### Does the Number of correlate with the Number of parents/Children ?

```python
print(Titanic[['SibSp', 'Survived']].groupby(['SibSp']).mean())
print('\n\n')
print(Titanic[['Parch', 'Survived']].groupby(['Parch']).mean())
```
